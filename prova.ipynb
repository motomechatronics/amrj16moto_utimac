# 5. File and Folder Management

I will now demonstrate how to create and manage files. This is often used to create log folders or temporary configuration backups.

* **Commands:** `mkdir` (create folder), `touch` (create file), `cp` (copy), `mv` (move/rename), `rm` (remove).

```python
# 1. Creating a directory for my personal logs
!mkdir my_logs

# 2. Creating an empty log file
!touch my_logs/session1.log

# 3. Creating a backup of the README file
!cp README.md README_backup.md

# 4. Renaming the backup (Moving it to the logs folder)
!mv README_backup.md my_logs/README_old.md

# 5. Verifying the result
!ls -R my_logs

# 6. Cleaning up (Deleting the folder)
!rm -r my_logs
```

**Personal Comment:**
I successfully created and moved files. I used the `-r` flag with `rm` (remove recursive) because `rm` alone cannot delete a directory that contains files. This is a dangerous command, so I double-checked the folder name before pressing enter.

# 7. Search

* **Command:** `grep`

# 8. Permissions

This is a critical section. In ROS 2, if a Python node script is not executable (`+x`), the ros2 run command will fail.

* **Command:** `chmod +x`
    * *Why:* To make a script executable.

```python
# Creating a dummy script
!echo "echo 'Hello Robot'" > test_robot.sh

# Checking default permissions (usually rw-rw-r--)
!ls -l test_robot.sh

# Changing mode to Executable
!chmod +x test_robot.sh

# Verifying the 'x' flag is present
!ls -l test_robot.sh

# Running it
!./test_robot.sh

# Cleanup
!rm test_robot.sh
```

**Personal Comment:**
I observed how the permission string changed from `-rw-rw-r--` to `-rwxrwxr-x`. The green color (in some terminals) confirms it is now an executable program. This step is often forgotten by beginners but is mandatory for robot scripts.
